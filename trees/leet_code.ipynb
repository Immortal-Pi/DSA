# min distance

```python
# Definition for a binary tree node.
from collections import deque
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def minDiffInBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        if root==None:
            return 0
        difference_array=[]
        queue=deque([root])
        ans_queue=[]
        min_val=float('inf')
        while queue:
            top=queue.popleft()
            for i in range(len(ans_queue)):
                min_val=min(abs(top.val-ans_queue[i]),min_val)
            ans_queue.append(top.val)
            if top.left:
                queue.append(top.left)
            if top.right:
                queue.append(top.right)
        return min_val
```

```python
# Definition for a binary tree node.
from collections import deque
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def minDiffInBST(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: int
        """
        self.ans=float('inf')
        self.pred=None
        self.inorder_traversal(root)
        return self.ans

    def inorder_traversal(self,root):
        if root==None:
            return 0
        self.inorder_traversal(root.left)
        if self.pred is not None:
            self.ans=min(self.ans,abs(root.val-self.pred))
        self.pred=root.val
        self.inorder_traversal(root.right)
        return self.ans


        
```

```python
# mirror 
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val=0, left=None, right=None):
#         self.val = val
#         self.left = left
#         self.right = right
class Solution(object):
    def isSymmetric(self, root):
        """
        :type root: Optional[TreeNode]
        :rtype: bool
        """
        if root==None:
            return True 
        return self.isMirror(root.left,root.right)

    def isMirror(self,t1,t2):
        if t1==None and t2==None:
            return True 
        if t1==None or t2==None:
            return False
        if t1.val!=t2.val:
            return False
        return self.isMirror(t1.left,t2.right) and self.isMirror(t1.right,t2.left)

        

        
```
