```python
class BinaryTree:

    def __init__(self,value):
        self.val=value
        self.left=None 
        self.right=None 
```

```python
root=BinaryTree(1)
root.left=BinaryTree(2)
root.right=BinaryTree(3)
root.left.left=BinaryTree(4)
root.left.right=BinaryTree(5)
```

```python
# print tree node 
def print_tree(root):
    if root==None:
        return root 

    print(root.val,end=':')
    if root.left!=None:
        print(root.left.val,end=',')
    else:
        print(f'None',end=',')
    if root.right!=None:
        print(root.right.val,end=',')
    else:
        print(f'None',end=',')
    print('\n')
    print_tree(root.left)
    print_tree(root.right)
    

print_tree(root)
```
