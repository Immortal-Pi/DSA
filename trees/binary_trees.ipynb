```python
class BinaryTree:

    def __init__(self,value):
        self.val=value
        self.left=None 
        self.right=None 
```

```python
root=BinaryTree(1)
root.left=BinaryTree(2)
root.right=BinaryTree(3)
root.left.left=BinaryTree(4)
root.left.right=BinaryTree(5)
```

```python
# print tree node 
def print_tree(root):
    if root==None:
        return root 

    print(root.val,end=':')
    if root.left!=None:
        print(root.left.val,end=',')
    else:
        print(f'None',end=',')
    if root.right!=None:
        print(root.right.val,end=',')
    else:
        print(f'None',end=',')
    print('\n')
    print_tree(root.left)
    print_tree(root.right)
    

print_tree(root)
```

# def take input for a tree

```python
def take_input_binary_tree():
    num=int(input('Enter the value of the node'))
    if num==-1:
        return None 
    root=BinaryTree(num)
    print(f'enter the left node val of {num}')
    root.left=take_input_binary_tree()
    print(f'enter the right value of {num}')
    root.right=take_input_binary_tree()

    return root
```

```python
root=take_input_binary_tree()
print_tree(root)
```

```python
# take input level wise
from collections import deque 
def take_input_binary_tree_lvl():
    num=int(input('enter the value of the Node'))
    root=BinaryTree(num)
    queue=deque([root])
    while queue:
        level=len(queue)
        top=queue.popleft()
        for _ in range(level):
            left=int(input(f'enter the left of the {top.val}'))
            top.left=BinaryTree(left)
            right=int(input(f'enter the right of the {top.val}'))
            top.right=BinaryTree(right)
            queue.append(top.left)
            queue.append(top.right)
    return top



root=take_input_binary_tree_lvl()
print_tree(root)
```
