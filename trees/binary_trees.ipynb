```python
class BinaryTree:

    def __init__(self,value):
        self.val=value
        self.left=None 
        self.right=None 
```

```python
root=BinaryTree(1)
root.left=BinaryTree(2)
root.right=BinaryTree(3)
root.left.left=BinaryTree(4)
root.left.right=BinaryTree(5)
```

```python
# print tree node 
def print_tree(root):
    if root==None:
        return root 

    print(root.val,end=':')
    if root.left!=None:
        print(root.left.val,end=',')
    else:
        print(f'None',end=',')
    if root.right!=None:
        print(root.right.val,end=',')
    else:
        print(f'None',end=',')
    print('\n')
    print_tree(root.left)
    print_tree(root.right)
    

print_tree(root)
```

# def take input for a tree

```python
def take_input_binary_tree():
    num=int(input('Enter the value of the node'))
    if num==-1:
        return None 
    root=BinaryTree(num)
    print(f'enter the left node val of {num}')
    root.left=take_input_binary_tree()
    print(f'enter the right value of {num}')
    root.right=take_input_binary_tree()

    return root
```

```python
root=take_input_binary_tree()
print_tree(root)
```

```python
# take input level wise
from collections import deque 
from binary_tree import BinaryTree,print_tree
def take_input_binary_tree_lvl():
    num=int(input('enter the value of the root'))
    root=BinaryTree(num)
    queue=deque([root])
    while queue:
        top=queue.popleft()
        left=int(input(f'enter the left of the {top.val}'))
        if left!=-1:
            left_node=BinaryTree(left)
            top.left=left_node
            queue.append(left_node)
        right=int(input(f'enter the right of the {top.val}'))
        if right!=-1:
            right_node=BinaryTree(right)
            top.right=right_node
            queue.append(right_node)
    return root 

root=take_input_binary_tree_lvl()
# print_tree(root)
```

# diameter of a tree 
- maximum distance between 2 points 

```python
def get_height(root):
    if root==None:
        return 0 
    left_height=get_height(root.left)
    right_height=get_height(root.right)
    heightOfTree=1+max(left_height,right_height)
    return heightOfTree
    
def get_diameter_of_BT(root):
    if root==None:
        return 0 
    left=get_height(root.left)
    right=get_height(root.right)

    left_diameter=get_diameter_of_BT(root.left)
    right_diameter=get_diameter_of_BT(root.right)
    num=max(left_diameter,right_diameter,left+right)
    return num 
```

# get diameter of a BT optimized way

```python
def get_diameter_of_BT_optimized(root):
    
    def get_height(node):
        if node==None:
            return 0,0 
        left_height,left_diameter=get_height(node.left)
        right_height,right_diameter=get_height(node.right)
        diameter_through_node=left_height+right_height
        height=1+max(left_height,right_height)
        diameter=max(diameter_through_node,left_diameter,right_diameter) 
        return height,diameter
    return get_height(root)[1]
```

# the given binary tree is balanced 
- | lh-rh| <=1 
- start with root if balanced go with sub trees and check for the same 


```python
   
def balanced_binary_tree(root):
    if root==None:
        return True
    def get_height(node):
        if node==None:
            return 0,True 
        left_height,balanced_left=get_height(node.left)
        right_height,balanced_right=get_height(node.right)
        if not balanced_left or not balanced_right:
                return 0,False
        current_height=max(left_height,right_height)+1
        balanced_height=True if abs(left_height-right_height)<=1 else False
        return current_height,balanced_height
    height,balance=get_height(root)
    return balance
```

# pre order traversal 


```python
def preorder_traversal(root):
    if root==None:
        return None
    print(root.val,end=' ')
    preorder_traversal(root.left)
    preorder_traversal(root.right) 
# print_tree_BT(root)
preorder_traversal(root)
```

```python
def postorder_traversal(root):
    if root==None:
        return 
    postorder_traversal(root.left)
    postorder_traversal(root.right)  
    print(root.val,end=' ')
postorder_traversal(root)
```

```python
def inorder_traversal(root):
    if root==None:
        return 
    inorder_traversal(root.left)
    print(root.val,end=' ')
    inorder_traversal(root.right)  
    
inorder_traversal(root)
```

```python
from collections import deque 
from binary_tree import BinaryTree,print_tree
def construct_tree_preorder_indorder(inorder,preorder,inS, inE, prS, prE):
    print(inorder,preorder,inS, inE, prS, prE)

    if inS>inE or prS>prE:
        return None
    rootValue=preorder[prS]
    root=BinaryTree(preorder[prS])
    inorder_element_index=-1
    for i in range(inS,inE+1):
        if inorder[i]==rootValue:
            inorder_element_index=i
            break
    left_size=inorder_element_index-inS
    root.left=construct_tree_preorder_indorder(inorder,preorder,inS,inorder_element_index-1,prS+1,prS+left_size)
    root.right=construct_tree_preorder_indorder(inorder,preorder,inorder_element_index+1,inE,prS+1+left_size,prE)
    return root
inorder=[4,2,5,1,3,6]
preorder=[1,2,4,5,3,6]
tree=construct_tree_preorder_indorder(inorder,preorder,0, len(inorder)-1,0, len(preorder)-1)
print_tree(tree)
```
