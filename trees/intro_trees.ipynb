```python
class TreeNode:
    def __init__(self,data):
        self.data=data
        self.child=[]

    def add_children(self,node):
        self.child.append(node)
```

```python
normal_root=TreeNode(1)
child1=TreeNode(2)
child2=TreeNode(3)
child3=TreeNode(4)
normal_root.child=[child1,child2,child3]
```

```python
# visual example 
organization=TreeNode('CEO')
head_of_sales=TreeNode('Head of Sales')
head_of_marketing=TreeNode('Head of Marketing')
sales_executive=TreeNode('Sales Executive')
marketing_executuve=TreeNode('Marketing Executive')
organization.child.append(head_of_sales)
organization.child.append(head_of_marketing)
organization.child.append(sales_executive)
organization.child.append(marketing_executuve)
```

```python
# print a tree 
def print_tree(tree):
    if tree==None: # edge case
        return 
    print(tree.data,end='->')
    print('\n')
    for i in range(len(tree.child)):
        print_tree(tree.child[i])

print_tree(organization)
```

```python
def print_detailed_tree(tree):
    if tree==None: # edge case
        return 
    print(tree.data,end='-:')
    for i in tree.child:
        print(i.data,end=',')
    print('\n')
    for i in range(len(tree.child)):
        print_detailed_tree(tree.child[i])
        

print_detailed_tree(organization)
```

```python
# take input level wise 
from treeStructure import TreeNode,test_cases
from collections import deque 

def take_input_level_wise():
    data=int(input(f'Enter the data for tree node'))
    root=TreeNode(data)
    queue=deque([root])

    while len(queue)!=0:
        current_node=queue.popleft()
        num_children=int(input(f'Enter the number of children for {current_node.data}'))
        for i in range(num_children):
            child_data=int(input(f'Enter the data for child node {current_node.data}'))
            child_node=TreeNode(child_data)
            current_node.child.append(child_node)
            queue.append(child_node)
    return root
```

```python
root=take_input_level_wise()
root.print_detailed_tree(root)
```

```python
def count_nodes(root):
    if root==None:
        return 0 
    num=1
    for eachchild in root.child:
        # print(eachchild.data)
        num=num+count_nodes(eachchild)
    return num 
```

```python
# height
def get_tree_height(root):
    if root==None:
        return 0
    height=1
    child_height=0
    for child in root.child:
        child_height=max(child_height,get_tree_height(child))
    return height+child_height

for name, tree_data in test_cases.items():
    print(f"\nTesting: {name}")

    if tree_data is None:
        root = None
    else:
        root = build_trees(tree_data)

    print(get_tree_height(root))
```

# traversal of tree 
- preorder traversal - root and then children 
- postorder traversal - children then root 

```python
def preorder_traveral(root):
    if root is None:
        return 
    print(root.data,end=" ")
    for child in root.child:
        preorder_traveral(child)
```

```python
def postorder_traversal(root):
    if root is None:
        return 
    for child in root.child:
        postorder_traversal(child)   
    print(root.data,end=" ")
     
    
```

# largest Value in each tree 

```python
from collections import deque
from treeTestCases import test_cases,build_tree,print_tree

def largest_values_in_rows(root):
    if root==None:
        return []

    queue=deque([root])
    result=[]
    while queue:
        max_size=len(queue)
        max_val=float('-inf')
        
        for _ in range(max_size):
            current=queue.popleft()
            max_val=max(max_val,current.val)

            for child in current.child:
                queue.append(child)
        result.append(max_val)
    return result

    
    

tree1=build_tree(test_cases['complex_tree'])
# print_tree(tree1)
largest_values_in_rows(tree1)
```
